```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Demo").getOrCreate()

print("Spark chal gaya 🚀")
```

```python
data = [
    ("Laptop", 50000),
    ("Mobile", 20000),
    ("Tablet", 15000),
    ("Laptop", 55000),
    ("Mobile", 22000),
    ("Tablet", 18000),
    ("Laptop", 60000)
]

columns = ["Product", "Price"]

df = spark.createDataFrame(data, columns)

df.show()
```

```python
df.count()
```

```python
from pyspark.sql.functions import sum

df.select(sum("Price")).show()
```

```python
df.groupBy("Product").sum("Price").show()
```

```python
df.groupBy("Product").count().show()
```

```python
import matplotlib.pyplot as plt

pdf = df.groupBy("Product").sum("Price").toPandas()

plt.bar(pdf["Product"], pdf["sum(Price)"])
plt.title("Product Wise Sales")
plt.xlabel("Product")
plt.ylabel("Sales Amount")
plt.show()
```

Conclusion:
We used PySpark to analyze product sales data.
Laptop generated the highest revenue.
This demonstrates scalable big data analysis using Spark.
